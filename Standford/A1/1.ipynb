```python
text = "low low low low low lower lower widest widest widest newest newest newest newest newest newest"
words = list(text.split(' '))
freq_words = {}
for i in words:
    if i not in freq_words:
        freq_words[i]=0
    freq_words[i]+=1
freq_words
```

```python
vocab = {}
for i in range(256):
    vocab[i] = bytes([i])
word_tokens = {}
for word, freq in freq_words.items():
    word_tokens[word] = list(word.encode("utf-8"))
```

```python
code = 256
for i in range(1):
    pair_count = {}
    for word, freq in freq_words.items():
        tokens = word_tokens[word]
        for j in range(len(tokens) - 1):
            pair = (tokens[j], tokens[j + 1])
            if pair not in pair_count:
                pair_count[pair] = 0
            pair_count[pair] += freq
    max_freq = max(pair_count.values())
    pair = None
    for pairs, freq in pair_count.items():
        if freq == max_freq:
            if pair is None or pair < pairs:
                pair = pairs
    vocab[code] = vocab[pair[0]] + vocab[pair[1]]
    print(pair, "->", code, vocab[code])
    for word in word_tokens:
        tokens = word_tokens[word]
        new_tokens = []
        j = 0
        while j < len(tokens):
            if j < len(tokens) - 1 and tokens[j] == pair[0] and tokens[j + 1] == pair[1]:
                new_tokens.append(code)
                j += 1
            else:
                new_tokens.append(tokens[j])
            j += 1
        word_tokens[word] = new_tokens
    code += 1
```
